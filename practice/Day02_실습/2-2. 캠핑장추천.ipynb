## 고캠핑 API 사용하여 캠핑 정보 가져오기 
- https://www.data.go.kr/data/15101933/openapi.do 고캠핑 API : api 활용신청하여 발급받아 .env 파일에 저장하기        

- 캠핑장 API 사용 : GET http://apis.data.go.kr/B551011/GoCamping/searchList       

- 반환 정보 예시
```python
    "items": {
    "item": [
        {
        "facltNm": "덕풍계곡 솔밭야영장",
        "intro": "산속에 위치한 가족형 캠핑장입니다."  ← 있는 경우

```

````python
# 환경설정 파일에서 서비스 키를 가져오는 코드
import os
ServiceKey = os.getenv("GOCAMPING_SERVICE_KEY")
````

````python
from urllib.parse import quote

# '야영장' 키워드 URL 인코딩
encoded_keyword = quote("야영장")
print(encoded_keyword)

# quote()는 URL에 넣을 수 없는 문자열(특히 한글)을 안전하게 변환(인코딩)하기 위해 사용
````

````python
# url 정의하기
base_url = "http://apis.data.go.kr/B551011/GoCamping"
url = f"{base_url}/searchList?serviceKey={ServiceKey}&keyword={encoded_keyword}&numOfRows=10&pageNo=1&MobileOS=ETC&MobileApp=TestApp&_type=json"
print(url)
````

````python
# url 호출 함수 정의하기
import requests
import json

def get_url_content(url):
    try:
        response = requests.get(url)
        if response.status_code == 200: # 성공
            return response.text 
            # json 형태의 캠핑장 text data
        else: # 실패시 
            response.raise_for_status()

    except requests.exceptions.RequestException as e:
        print(f"Request Exception: {e}")
        return None

# url을 통해 API 호출하기
result = get_url_content(url)

if result is not None:
    print(result)
    #print(json.dumps(result, indent=4, ensure_ascii=False)) 
else:
    print("Failed to retrieve URL content.")
````

````python
import json

# API가 준 JSON을 dict로 변환
data = json.loads(result)
sites = data['response']['body']['items']['item'] # 실제 캠핑장 list를 꺼내 site에 저장
sites
````

````python
# sites 출력하기
for site in sites:
    print(site['facltNm'] + ":" + site['intro'])
````

## 캠핑 정보 검색한 결과를 바탕으로 GPT 추천 서비스
1. 키워드 추출
2. 공공데이터 포털 검색 URL 생성
3. 검색 결과 파싱 및 캠핑장 리스트 생성
4. 캠핑장 리스트에 대해 GPT로 분석 요청

````python
# 1. 필요한 라이브러리 임포트 및 환경 변수 설정
import os
import json
import requests
from openai import OpenAI
from dotenv import load_dotenv
from urllib.parse import quote  # 한글 등 인코딩을 위한 quote 함수

# 2. OpenAI 클라이언트 초기화
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")
service_key = os.getenv("GOCAMPING_SERVICE_KEY")

client = OpenAI(api_key=api_key)
````

````python
# 3. URL 요청 함수 정의
def get_url_contet(url):
    try:
        response = requests.get(url)
        if response.status_code == 200:
            return response.text
        else:
            response.raise_for_status()
    except requests.exceptions.RequestException as e:
        print(f"요청 오류: {e}")
        return None
````

````python
# 4. GPT를 이용해 질문에서 핵심 검색어만 추출하는 함수 정의
def extract_keyword_from_question(question):
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[
            {
                "role": "system",
                "content": """
너는 캠핑장 검색 키워드 추출기다.

사용자의 질문에서 공공 캠핑장 API 검색에 사용할 핵심 키워드만 뽑아라.

규칙:
1. 질문에 실제로 등장한 단어만 사용한다.
2. 질문에 없는 지역명, 장소명은 절대 만들지 않는다.
3. '캠핑장', '추천', '할 수 있는', '있는' 같은 일반 단어는 제거한다.
4. 출력은 검색 키워드만 한다.
5. 설명하지 않는다.

예시:
질문: 계곡이 있는 캠핑장을 추천해줘.
출력: 계곡

질문: 바베큐를 할 수 있는 캠핑장을 추천해줘.
출력: 바베큐

질문: 강원도에 있는 캠핑장 추천해줘.
출력: 강원도

질문: 서울 근처 글램핑 추천해줘.
출력: 서울 글램핑
"""
            },
            {
                "role": "user",
                "content": question
            }
        ]
    )
    return response.choices[0].message.content.strip()
````

````python
def extract_sites_from_result(result):
    if result is None:
        return []
    
    data = json.loads(result)

    body = data.get("response", {}).get("body", {})
    items = body.get("items", "")

    # 검색 결과가 없으면 itmes가 ""로 오는 경우가 있음
    if items == "" or items is None:
        return []
    
    sites = items.get("item", [])

    # 결과가 1개일 때 dict로 오는 경우가 있으므로 list로 변환
    if isinstance(sites, dict):
        sites = [sites]

    return sites
````

````python
# 6. 질문 입력 → 키워드 추출 → 공공데이터 API 호출
question = "바베큐를 할 수 있는 캠핑장을 원해."

keyword = extract_keyword_from_question(question)
keyword = keyword.replace(":", " ") 

encoded_keyword = quote(keyword)

print(f"추출된 키워드: {keyword}")
print(f"URL-safe 인코딩된 키워드: {encoded_keyword}")

url = f"http://apis.data.go.kr/B551011/GoCamping/searchList?serviceKey={service_key}&keyword={encoded_keyword}&numOfRows=10&pageNo=1&MobileOS=ETC&MobileApp=TestApp&_type=json"

result = get_url_content(url)
````

````python
# 7. JSON 결과 파싱하여 캠핑장 리스트 추출
sites = extract_sites_from_result(result)

# 8. 캠핑장 정보를 번호와 함께 문자열로 정리
context = ""

for i, site in enumerate(sites, 1):
    name = site.get("facltNm", "이름 없음")
    intro = site.get("intro") or site.get("lineIntro") or site.get("featureNm") or "설명 없음"
    address = site.get("addr1", "")
    category = site.get("induty", "")

    context += f"{i}) 이름: {name}\n"
    context += f"   소개: {intro}\n"
    context += f"   주소: {address}\n"
    context += f"   유형: {category}\n\n"

print(f"검색된 캠핑장 수: {len(sites)}")
print("📄 캠핑장 목록:\n")
print(context)
    

````

````python
# 9. GPT에게 캠핑장 리스트를 기반으로 질문에 맞는 추천 요청
if len(sites) == 0:
    print("검색된 캠핑장이 없습니다. 다른 키워드로 다시 검색해보세요.")
else:
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[
            {
                "role": "system",
                "content": """
너는 캠핑장 추천 도우미다.
사용자의 질문과 공공데이터 api에서 가져온 캠핑장 목록을 보고, 
질문 조건에 가장 잘 맞는 캠핑장을 2~3개 추천해라.audio=

답변 형식:
1. 추천 캠핑장명
- 추천 이유:
- 위치:
- 주의할 점:

목록에 없는 캠핑장은 절대 지어내지 마라.
"""
            },
            {
                "role": "user",
                "content": f"사용자 질문: {question}\n\n캠핑장 목록: \n{context}"
            }
        ] 
    )

    print("GPT 추천 결과: ")
    print(response.choices[0].message.content)
````
